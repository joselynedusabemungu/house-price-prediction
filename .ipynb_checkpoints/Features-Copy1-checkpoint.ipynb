```python
from sklearn.preprocessing import LabelEncoder

train_label = pd.DataFrame()
label_encoder = LabelEncoder()

for col in X.columns:
    if X[col].dtype == 'object':  
        train_label[col] = label_encoder.fit_transform(X[col].astype(str))
    else:
        train_label[col] = X[col]

print('Train data has {} rows and {} columns'.format(train_label.shape[0], train_label.shape[1]))
train_label.head(5)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.impute import SimpleImputer

def train_model(X_data, Y_data):
    # 1. Split the data first to prevent data leakage
    X_train, X_test, Y_train, Y_test = train_test_split(X_data, Y_data, random_state=42, test_size=0.2)
    
    # 2. Fill missing (NaN) values using the mean of each column
    imputer = SimpleImputer(strategy='mean')
    X_train = imputer.fit_transform(X_train)
    X_test = imputer.transform(X_test) 
    
    model = LogisticRegression(max_iter=100000)
    model.fit(X_train, Y_train)
    
    predictions = model.predict(X_test)
    print('Accuracy: {:.4f}'.format(accuracy_score(Y_test, predictions)))
    return model

print("Label Encoding Accuracy:", accuracy_score(Y_test, predictions))
train_model(Y_test, Y_train)
```
